```python
import re
import json

def parse_showdown_items(js_text: str) -> dict:
    # 1. Strip the JS assignment prefix and trailing semicolon
    js_text = js_text.strip()
    js_text = re.sub(r"^exports\.BattleItems\s*=\s*", "", js_text)
    js_text = js_text.rstrip(";")

    # 2. Quote unquoted keys: turn  abc:{  into  "abc":{
    #    Regex: start of line or { or comma, then identifier, then colon
    js_text = re.sub(r'([{,])\s*([A-Za-z0-9_]+)\s*:', r'\1 "\2":', js_text)

    # 3. The result should now be valid JSON
    try:
        return json.loads(js_text)
    except json.JSONDecodeError as e:
        # If there's a decode error, print the offending snippet for debugging
        print("JSON decode error:", e)
        raise
```

```python
with open("abilities.txt", "r") as f:
    items = parse_showdown_items(f.read())
```

```python
print(json.dumps(items, indent=2))
```

```python
from poke_env.player import Player, RandomPlayer

class DebugPlayer(Player):
    def choose_move(self, battle):
        print("=== New turn ===")
        try:
            # for attr in attributes:
            #     print(f'{attr} ({type(battle.active_pokemon.__getattribute__(attr))}): {battle.active_pokemon.__getattribute__(attr)}')
            print(battle.active_pokemon.ability)
            print(battle.opponent_active_pokemon)
        except:
            pass

        # For now, just act randomly
        return self.choose_random_move(battle)
```

```python
async def main():
    p1 = DebugPlayer(battle_format="gen9randombattle")
    p2 = RandomPlayer(battle_format="gen9randombattle")

    await p1.battle_against(p2, n_battles=1)

    print("Finished battle!")
    print("Wins:", p1.n_won_battles)

await main()
```
